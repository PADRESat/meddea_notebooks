```python
from roentgen.absorption import Material
import matplotlib.pyplot as plt
import astropy.units as u
from astropy.io import ascii
from astropy.time import Time
from specutils import Spectrum1D
from astropy.visualization import quantity_support
import numpy as np
quantity_support()
from pathlib import Path
```

```python
from padre_meddea.calibration.ea import get_effective_area
```

```python
energy = np.linspace(5, 100, 100) * u.keV
eff_area = get_effective_area(energy)
eff_area_nopath = get_effective_area(energy, no_al = True, no_be = True, no_mli = True)
eff_area_noal = get_effective_area(energy, no_al = True)
```

```python
plt.plot(energy, eff_area)
plt.plot(energy, eff_area_nopath, label="No filters")
plt.plot(energy, eff_area_noal, label="Be 1 mm, MLI")
plt.xlabel("Energy (keV)")
plt.ylabel("Effective Area [cm$^2$]")
plt.legend()
```
